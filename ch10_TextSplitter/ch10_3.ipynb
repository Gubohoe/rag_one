```python
import warnings
from langchain_text_splitters import (
    CharacterTextSplitter,
    TokenTextSplitter,
    SpacyTextSplitter,
    SentenceTransformersTokenTextSplitter,
    NLTKTextSplitter,
    )
from transformers import GPT2TokenizerFast
```

```python
# !uv add nltk
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
with open("../data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()
```

```python
print(file[:500])
```

```python
text_splitter = CharacterTextSplitter.from_tiktoken_encoder(
    chunk_size=300,
    chunk_overlap=0,
)
texts = text_splitter.split_text(file)
```

```python
print(len(texts))
```

```python
text_splitter2 = TokenTextSplitter(
    chunk_size=300,
    chunk_overlap=0,
)

texts2 = text_splitter2.split_text(file)
print(texts2[0])
```

```python
warnings.filterwarnings("ignore")

text_splitter3 = SpacyTextSplitter(
    chunk_size=200,
    chunk_overlap=50,
)

texts3 = text_splitter3.split_text(file)
print(texts3[0])
```

```python
splitter = SentenceTransformersTokenTextSplitter(chunk_size=200, chunk_overlap=0)

count_start_and_stop_tokens = 2

text_token_count = splitter.count_tokens(text=file) - count_start_and_stop_tokens
print(text_token_count)

text_chunks = splitter.split_text(text=file)
print(text_chunks[1])
```

```python
text_splitter4 = NLTKTextSplitter(
    chunk_size=200,
    chunk_overlap=0,
)

texts4 = text_splitter4.split_text(file)
print(texts4[0])
```

```python
hf_tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")

text_splitter5 = CharacterTextSplitter.from_huggingface_tokenizer(
    hf_tokenizer,
    chunk_size=300,
    chunk_overlap=50,
)

texts5 = text_splitter5.split_text(file)
print(texts5[1])
```
